```python
def happy_learner(review):
    if isinstance(review, str):
        if 'great' in review.lower():
            return 'Very Happy'
        else:
            return 'SOmething'
    else:
        return 'Some issue with your data'
```

```python
data['happy_learner'] = data['subjective_feedback'].apply(happy_learner)
```

```python
output = []

for review in data['subjective_feedback']:
    output.append(happy_learner(review))
    
data['happy_learner'] = output
```
