```python
email_body = [
    "Hi team, I am facing some issue while login into the application",
    "I have paid but facing issue at time of generating the invoice",
    "Not able to pay, some issue at time of billing",
    "I am getting authentication issue even my credentials are correct"
]
```

```python
def predict(text):
    billing=['billing','invoice',"server","payment"]
    technical=['error','authentication','login']

    for word in billing:
        if word in text:
            return "Billing"
    for word in technical:
        if word in text:
            return "Technical"
    return "Other"
```

```python
predict("I am not happpy with your product, none of module is working")
```

```python
from openai import OpenAI
from dotenv import load_dotenv
load_dotenv()
```

```python
client = OpenAI()
```

```python
def call_llm(email):
    response =client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
        
            "role": "user",
            "content": "Please classify my email as spam or not: It should not have any descriptions, thoery, just need category "+email
        }
    ]
)
    return response.choices[0].message.content
```

```python
call_llm('Please classify my email as spam or not: "Hi team, I am facing some issue while login into the application"')
```
